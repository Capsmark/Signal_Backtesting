```python
import pandas as pd

# Load the dataset
file_path = 'E:\Signal Backtesting\Output\\178_tp=0.95_sl=1.6.csv'
df = pd.read_csv(file_path)
```

```python
# Convert 'Datetime' and 'Execution Duration' columns to datetime objects
df['Datetime'] = pd.to_datetime(df['Datetime'])
df['Execution Duration'] = pd.to_timedelta(df['Execution Duration'])

# Add 'Execution Duration' to 'Datetime' to create a new datetime
df['New Datetime'] = df['Datetime'] + df['Execution Duration']

# Map 'Side' to 'Signal'
df['Signal'] = df['Side'].map({'Sell': -2, 'Buy': 2})

# Create a new dataframe with the required columns
new_df = df[['New Datetime', 'Signal']].rename(columns={'New Datetime': 'Datetime'})

# Save the new dataframe to a CSV file
output_file_path = 'E:\Signal Backtesting\Output\\178\\updated_dataset.csv'
new_df.to_csv(output_file_path, index=False)


```
