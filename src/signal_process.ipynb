```python
import pandas as pd

# Load the initial CSV file and parse 'Datetime'
df = pd.read_csv('E:\Signal Backtesting\Input\\updated_sigai_signal_dataset_with_more_important_events.csv', parse_dates=['Datetime'])

# Convert 'Datetime' to datetime and add one hour
df['Datetime'] = pd.to_datetime(df['Datetime']) + pd.Timedelta(hours=1)

# Initialize variables for signal filtering
filtered_signals = []
seen_signal = False

# Iterate through each row in the signal data
for i, row in df.iterrows():
    signal_datetime = row['Datetime']
    signal_value = row['Signal']

    if signal_value == 0:
        seen_signal = False
        filtered_signals.append(row)
        continue

    if not seen_signal:
        if signal_value == 1 or signal_value == -1:
            seen_signal = True
            filtered_signals.append(row)
    else:
        row['Signal'] = 0
        filtered_signals.append(row)

# Convert the filtered signals to a DataFrame
filtered_signals_df = pd.DataFrame(filtered_signals)

# Save the filtered signals to a final CSV file
final_output_file_path = 'E:\\Signal Backtesting\\Input\\filtered_signals_with_most_important_events_plus_one_hour.csv'
filtered_signals_df.to_csv(final_output_file_path, index=False)
```
