```python
import pandas as pd

# Load the datasets
file_path1 = 'E:\SignalModel\Out_7.csv'
data = pd.read_csv(file_path1)
```

```python
# Filter out rows for the year 2023 only
data['datetime'] = pd.to_datetime(data['datetime'])
data_2023 = data[data['datetime'].dt.year == 2023]

# Initialize an empty list to store the new dataframe rows
new_rows = []

# Iterate over the rows and process the signals
previous_signal = None

for index, row in data_2023.iterrows():
    current_datetime = row['datetime']
    if row['SS'] == 1:
        if previous_signal != 'SS':
            new_rows.append({'Datetime': current_datetime, 'Signal': -2})
            previous_signal = 'SS'
    elif row['SB'] == 1:
        if previous_signal != 'SB':
            new_rows.append({'Datetime': current_datetime, 'Signal': 2})
            previous_signal = 'SB'
    else:
        previous_signal = None

# Create a new dataframe from the list of new rows
new_df = pd.DataFrame(new_rows)


new_df.to_csv('E:\Signal Backtesting\Output\\2023_signal.csv')
```

```python
# Load the provided signal data
file_path = 'E:\Signal Backtesting\Output\\2023_signal.csv'
signal_data = pd.read_csv(file_path, parse_dates=['Datetime'])

# Add 2:58:00 to the datetimes and write in 'Datetime 1' column
signal_data['Datetime '] = signal_data['Datetime'] + pd.Timedelta(hours=2, minutes=58)
signal_data.to_csv('E:\Signal Backtesting\Output\\2023_signal.csv')
```
